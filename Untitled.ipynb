```python
import pandas as pd
import numpy as np

# 1. 압축된 큰 파일 불러오기
# (파일명을 본인 파일명으로 수정하세요)
filename = "robotics_papers.csv.gz" 
print("파일 로딩 중...")
df = pd.read_csv(filename)

# 2. 반으로 쪼개기
mid_index = len(df) // 2
df_part1 = df.iloc[:mid_index]
df_part2 = df.iloc[mid_index:]

# 3. 각각 압축해서 저장하기
print("분할 저장 중...")
df_part1.to_csv("arxiv_data_part1.csv.gz", index=False, compression='gzip')
df_part2.to_csv("arxiv_data_part2.csv.gz", index=False, compression='gzip')

print("완료! part1과 part2 파일이 생성되었습니다.")
```

```python
unique_categories = (
    df['Category']
    .dropna()                      # NaN 제거
    .str.split(',')                # 콤마 기준 분리
    .explode()                     # 리스트 → 행으로 펼치기
    .str.strip()                   # 앞뒤 공백 제거
    .unique()                      # 유니크 값
)

list(unique_categories)
```
